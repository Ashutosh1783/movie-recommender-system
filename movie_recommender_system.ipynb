```python
import ast
import pandas as pd

def convert(obj):
    # If the object is already a list, return it as is.
    # This handles cases where the column has already been processed.
    if isinstance(obj, list):
        return obj

    # Handle NaN values or non-string types before literal_eval
    if pd.isna(obj) or not isinstance(obj, str):
        return []

    L = []
    try:
        # Attempt to evaluate the string as a Python literal (list of dicts)
        for i in ast.literal_eval(obj):
            L.append(i['name'])
    except (ValueError, SyntaxError):
        # Catch errors if the string is not a valid literal or malformed
        return []
    return L
```

```python
def convert3(obj):
    L = []
    counter=0
    for i in ast.literal_eval(obj):
        if counter != 3:
            L.append(i['name'])
            counter+=1
        else:
            break
    return L
```

```python
import pandas as pd

# Re-load movies and credits dataframes to ensure a clean state
m_df = pd.read_csv('/content/tmdb_5000_movies.csv')
c_df = pd.read_csv('/content/tmdb_5000_credits.csv')

# Rename 'id' column in movies_df to 'movie_id' to align with credits_df
m_df.rename(columns={'id': 'movie_id'}, inplace=True)

# Merge the dataframes on the common 'movie_id'
movies = m_df.merge(c_df, on='movie_id')

# Drop the redundant 'title_y' column and rename 'title_x' to 'title'
movies.drop(columns=['title_y'], inplace=True)
movies.rename(columns={'title_x': 'title'}, inplace=True)

# Now select the desired columns
movies=movies[['movie_id','title','overview','genres','keywords','cast','crew','tagline']]
```

```python
def fetch_director(obj):
    L = []
    for i in ast  .literal_eval(obj):
        if i['job'] == 'Director':
          L.append(i['name'])
          break
    return L

```

```python
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
def stem(text):
  y=[]
  for i in text.split():
    y.append(ps.stem(i))

  return " ".join(y)
```

```python
new_df['tags']=new_df['tags'].apply(lambda x: " ".join(x)).apply(stem)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity=cosine_similarity(vectors)
```

```python
sorted(list(enumerate(similarity[0])),reverse=True,key=lambda x:x[1])[1:6]
```

```python
print(similarity[0])
```

```python
def recommend(movie):
    movie_index = new_df[new_df['title'] == movie].index[0]
    distances = similarity[movie_index]

    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]

    for i in movies_list:
        print(new_df.iloc[i[0]].title)

# function call
recommend('Batman Begins')
```
